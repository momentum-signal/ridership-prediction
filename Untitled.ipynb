```python
import pandas as pd
import numpy as np

# Assume df is the original DataFrame with origin, destination, date, time, etc.
# Combine origin and destination into a route ID
df = pd.read_csv("/Users/nik/github/ridership-prediction/model/data/cleaned_data.csv")
df['route'] = df['origin'].astype(str) + "_" + df['destination'].astype(str)
# Sort by route and date/time to ensure correct ordering
df["ridership"] = df["ridership"].astype(float)
df["time_float"] = pd.to_datetime(df["time"], format="%H:%M").dt.hour + pd.to_datetime(df["time"], format="%H:%M").dt.minute / 60
df['dayofweek_sin'] = np.sin(2 * np.pi * (df['day_of_week'] - 1) / 7)
df['dayofweek_cos'] = np.cos(2 * np.pi * (df['day_of_week'] - 1) / 7)

df = df.sort_values(['route', 'date', 'time']).reset_index(drop=True)
# Assign a time index within each route
df['time_idx'] = df.groupby('route').cumcount()

# Ensure each route has at least 8 points (train+7)
min_length = 12
group_sizes = df.groupby('route').size()
valid_routes = group_sizes[group_sizes >= min_length].index
df = df[df['route'].isin(valid_routes)].copy()

# Determine cutoff per route: keep x < max_idx-6 for training, last 7 for validation
df['max_idx'] = df.groupby('route')['time_idx'].transform('max')
df['is_train'] = df['time_idx'] < (df['max_idx'] - 6)
train_df = df[df['is_train']].copy()
val_df   = df[~df['is_train']].copy()
```

```python
from pytorch_forecasting import TimeSeriesDataSet
from pytorch_forecasting.data import GroupNormalizer

# Define maximum encoder length (use full history for each route) 
max_encoder_length = train_df['time_idx'].max()  # or any suitable length

# Create TimeSeriesDataSet for training
training = TimeSeriesDataSet(
    train_df,
    time_idx="time_idx",
    target="ridership",
    group_ids=["route"],
    min_encoder_length=1,
    max_encoder_length=30,
    min_prediction_length=1,
    max_prediction_length=7,
    static_categoricals=["route"],
    static_reals=[],
    time_varying_known_categoricals=[],
    time_varying_known_reals=["time_float", "dayofweek_sin", "dayofweek_cos", "is_weekend", "is_holiday"],
    time_varying_unknown_categoricals=[],
    time_varying_unknown_reals=["ridership"],
    target_normalizer=GroupNormalizer(groups=["route"], transformation="softplus"),
    add_relative_time_idx=True,
    add_target_scales=True,
    add_encoder_length=True,
)
```

```python
import lightning.pytorch as pl
from pytorch_forecasting import TemporalFusionTransformer
from pytorch_forecasting.metrics import RMSE

# DataLoaders
batch_size = 32
train_dataloader = training.to_dataloader(train=True, batch_size=64, persistent_workers=False, num_workers=0)
# For validation, use the full DataFrame with predict=True to include last 7 points per route
validation = TimeSeriesDataSet.from_dataset(training, df, predict=True, stop_randomization=True)
val_dataloader = validation.to_dataloader(train=False, batch_size=64, persistent_workers=False, num_workers=0)

# Instantiate TFT model
tft = TemporalFusionTransformer.from_dataset(
    training,
    learning_rate=0.03,
    hidden_size=16,
    hidden_continuous_size=8,
    output_size=1,            # one output per time step (ridership)
    loss=RMSE(),              # point-loss (RMSE) for single-target regression
    dropout=0.1,
    attention_head_size=4,
    optimizer="adam",
    reduce_on_plateau_patience=4,
)
print(f"Number of parameters in network: {tft.size()/1e3:.1f}k")
for batch in train_dataloader:
    print("Batch found ✅")
    break
# Train the model
trainer = pl.Trainer(max_epochs=10, enable_model_summary=True, gradient_clip_val=0.1)
trainer.fit(tft, train_dataloaders=train_dataloader, val_dataloaders=val_dataloader)
```

```python
# Generate predictions (output and actual targets) on the validation set
predictions = tft.predict(val_dataloader, mode="prediction", return_x=False)
# predictions.output and predictions.y are tensors of shape (batch_size, 7)

# Example: print first batch of predicted vs actual ridership
pred_array = predictions.output.detach().cpu().numpy().flatten()
actual_array = predictions.y.detach().cpu().numpy().flatten()
print("Predicted vs Actual ridership (first examples):")
for p, a in zip(pred_array[:20], actual_array[:20]):
    print(f"Pred: {p:.2f}, Actual: {a:.2f}")
```
