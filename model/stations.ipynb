```python
import csv
import json
from collections import defaultdict
```

#### Load the CSV file

```python
csv_file = './data/cleaned_data.csv'
output_json = '../frontend/public/stations.json'
```

#### Process CSV and exclude self-mapping

```python
origin_to_dest = defaultdict(set)

with open(csv_file, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        origin = row['origin'].strip()
        destination = row['destination'].strip()
        if origin != destination:
            origin_to_dest[origin].add(destination)

# Convert sets to sorted lists
optimal_json = {origin: sorted(destinations) for origin, destinations in origin_to_dest.items()}

# Save to JSON file
with open(output_json, 'w', encoding='utf-8') as f:
    json.dump(optimal_json, f, indent=4, ensure_ascii=False)

# Print one specific origin's destinations
station = "Abdullah Hukum"
print(f'"{station}": {json.dumps(optimal_json.get(station, []), ensure_ascii=False, indent=4)}')
```
